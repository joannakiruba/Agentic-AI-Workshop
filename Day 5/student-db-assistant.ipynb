````python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model = "gemini-3.5-flash-lite",
    temperature = 0
)
````

````python
from langchain.tools import tool
import sqlite3


@tool
def get_student_info(student_id: str) -> dict:
    """
    Fetch basic student information.
    """

    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()

    cursor.execute(
        """
        SELECT student_id, name, department
        FROM students
        WHERE student_id = ?
        """,
        (student_id,)
    )

    row = cursor.fetchone()

    conn.close()

    if row is None:
        return {
            "error": "unknown_student"
        }

    return {
        "student_id": row[0],
        "name": row[1],
        "department": row[2]
    }
````

````python
@tool
def get_student_marks(student_id: str) -> dict:
    """
    Fetch marks of a student.
    """

    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()

    cursor.execute(
        """
        SELECT python, database, ai, web
        FROM students
        WHERE student_id = ?
        """,
        (student_id,)
    )

    row = cursor.fetchone()

    conn.close()

    if row is None:
        return {
            "error": "unknown_student"
        }

    return {
        "python": row[0],
        "database": row[1],
        "ai": row[2],
        "web": row[3]
    }
````

````python
@tool
def calculator(expression: str) -> dict:
    """
    Evaluate a mathematical expression.
    Used for totals and averages.
    """

    try:
        result = eval(expression)

        return {
            "result": result
        }

    except Exception:
        return {
            "error": "invalid_expression"
        }
````

````python
@tool
def get_passing_rules() -> dict:
    """
    Return university passing rules.
    """

    return {
        "minimum_average": 40,
        "minimum_subject_mark": 35
    }
````

````python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[get_student_info, get_student_marks, calculator, get_passing_rules],
    system_prompt="""
    Available tools:
    1. get_student_info
    2. get_student_marks
    3. calculator
    4. get_passing_rules

    Use the appropriate tool(s) to answer.
    """
)
````

````python
question = input("Ask agent: ")

result = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": question
        }
    ]
})

tool_chain = [
    msg.name
    for msg in result["messages"]
    if type(msg).__name__ == "ToolMessage"
]

print("\n"+"="*60)
print("Tool Chain:", " -> ".join(tool_chain))
print("="*60)

answer = result["messages"][-1].content

if isinstance(answer, list):
  answer = "\n".join(
      item["text"]
      for item in answer
      if isinstance(item, dict) and item.get("type") == "text"
  )

print("\n"+"="*60)
print("STUDENT-DB AGENT ANSWER")
print("="*60)
print(answer)
print("="*60)
````
